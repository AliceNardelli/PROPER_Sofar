```python
import os
import json
import openai
import ast
from omegaconf import OmegaConf
file_path_openai_config = r'C:\Users\alice\OneDrive\Desktop\EI_test_openai\config_openai_EI.yaml'
openai_config = OmegaConf.load(file_path_openai_config).config
openai.organization = openai_config.organization_id
openai.api_key = openai_config.api_key
system_message = openai_config.system_message_A2
```

```python
def run_completion( message):
    messages = [
        {"role": "system", "content": system_message},
    ]
    # Add each new message to the list
    messages.append({"role": "user", "content": message})

    # Request gpt-3.5-turbo for chat completion
    response = openai.ChatCompletion.create(
        model="gpt-4",
        messages=messages, 
        #max_tokens=max_tokens
        
    )

    # Print the response and add it to the messages list
    chat_message = response['choices'][0]['message']['content']
   
    messages.append({"role": "assistant", "content": chat_message})
    #chat_message = json.loads(chat_message)
    print(chat_message)
    return chat_message
```

```python
Emotions=["Happy","Neutral","Sad", "Angry","Fear","Surprise", "Disgust"]
#Emotions=["Happy"]
for e in Emotions:
    outputs=[]
    dir=r'C:\Users\alice\OneDrive\Desktop\EI_test_openai\generated_input'
    file_em=e+'.txt'
    path=os.path.join(dir,file_em )

    # Read the lines from the file
    file1 = open(path, 'r')

    # Print each line
    count = 0
    while True:
        line = file1.readline()
        
        try:
            ll=line.replace(",\n","")
            pp=json.loads(ll)
            new_dict={"text":pp["text"],"user_emotion": pp["user_emotion"],"comfortability": pp["comfortability"]}
            print(new_dict)
            line=str(new_dict)
            print(line)
            a=run_completion(line)
            print(a)
            outputs.append(a)
        except:
            print("void line")
            break
        if not line:
            break
        count += 1
        #print(f"Line {count}: {line.strip()}")
    file1.close()
    dir_res=r'C:\Users\alice\OneDrive\Desktop\EI_test_openai\A_results'
    path_res=os.path.join(dir_res,file_em)
    with open(path_res, 'w+') as f:
        f.writelines(f"{item}\n" for item in outputs)
```
