```python
import matplotlib.pyplot as plt

values = [
    8.0, 7.0, 9.0, 8.0, 10.0, 7.0, 9.0, 6.0, 6.0, 5.0,
    7.0, 8.0, 10.0, 7.0, 7.0, 9.0, 8.0, 10.0, 7.0, 9.0,
    6.0, 8.0, 9.0
]

labels = [
    "REACT_HAPPY_EMOTION", "GREET A0", "REACT_HAPPY_EMOTION", "ASK_MAIN_INFO A0", "REACT_HAPPY_EMOTION",
    "ANSWER_WITH_A_QUESTION", "REACT_HAPPY_EMOTION", "ANSWER_WITH_A_QUESTION", "REACT_NEUTRAL_EMOTION", 
    "ASK_PREFERRED_ACTIVITIES A0", "REACT_HAPPY_EMOTION", "ANSWER", "REACT_HAPPY_EMOTION", "ANSWER_WITH_A_QUESTION", 
    "WAIT2 A0", "REACT_HAPPY_EMOTION", "SUGGEST_ACTIVITIES A0", "REACT_CALM_EMOTION", "ANSWER_WITH_A_QUESTION", 
    "REACT_HAPPY_EMOTION", "ANSWER_WITH_A_QUESTION", "REACT_HAPPY_EMOTION", "ANSWER"
]

# Plot
plt.figure(figsize=(10, 5))
plt.plot(range(len(values)), values, marker='o', color='b', linestyle='-', markersize=6)
plt.xticks(range(len(labels)), labels, rotation=90)  # Set the x-ticks to the labels
plt.xlabel('Categories')
plt.ylabel('Score')
plt.title('Line Plot of Values and Categories')
plt.tight_layout()
plt.show()

```
