```python
import requests
```

```python
HOST = '127.0.0.1'
PORT = '2013'
URL_PERCEPTION = f'http://{HOST}:{PORT}/api/post_perception'
URL_PERSONALITY = f'http://{HOST}:{PORT}/api/post_personality'
URL_EM = f'http://{HOST}:{PORT}/api/get_em_description'
```

```python
class ClientPerception:

    def post_perception(self, user, emotion, sentence, episodes):
        payload = {
            'user': user,
            'emotion': emotion,
            'sentence': sentence,
            'episodes': episodes
        }
        try:
            response = requests.post(URL_PERCEPTION, json=payload, timeout=10)
            print("Server response status:", response.status_code)
            if response.text:
                print("Server response body:", response.text)
            else:
                print("Server returned no content.")
        except requests.exceptions.RequestException as e:
            print("Request failed:", e)

    def post_personality(self, personality):
        payload = {
            'personality': personality,
        }
        
        try:
            response = requests.post(URL_PERSONALITY, json=payload, timeout=5)
            print("Server response status:", response.status_code)
            if response.text:
                print("Server response body:", response.text)
            else:
                print("Server returned no content.")
        except requests.exceptions.RequestException as e:
            print("Request failed:", e)

    
    def get_em(self):
        response = requests.get(URL_EM, timeout=5)
        if response.status_code == 200:
            payload = response.json()
            return payload['em_descrption']
        else:
            print(f"Error: {response.status_code}")
            return None
```

```python
client = ClientPerception()


```

```python
client.post_personality(
    personality="Extrovert and Agreeable"
)
```

```python
res = client.get_em()

print(res)
```
