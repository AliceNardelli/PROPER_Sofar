```python
import requests
```

```python
HOST = '127.0.0.1'
PORT = '2013'
URL = f'http://{HOST}:{PORT}/api/post_perception'
```

```python
class ClientPerception:

    def post_perception(self, user, emotion, sentence, episodes):
        payload = {
            'user': user,
            'emotion': emotion,
            'sentence': sentence,
            'episodes': episodes
        }
        try:
            response = requests.post(URL, json=payload, timeout=5)
            print("Server response status:", response.status_code)
            if response.text:
                print("Server response body:", response.text)
            else:
                print("Server returned no content.")
        except requests.exceptions.RequestException as e:
            print("Request failed:", e)
```

```python
client = ClientPerception()

# Example test request
client.post_perception(
    user="Alice",
    emotion="Anger",
    sentence="I had a great day!",
    episodes="Angry to be annoyed with question"
)
```
